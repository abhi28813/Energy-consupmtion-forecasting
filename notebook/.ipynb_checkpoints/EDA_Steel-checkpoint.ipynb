```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd
df = pd.read_csv('/Users/abhijithk/Documents/data project/Core 1/Steel_industry_data.csv')
df.sample(4)
```

```python
df.info()
```

```python
df['Date'] = df['date'].str.split(' ').str[0]

df['Time'] = df['date'].str.split(' ').str[1]
```

```python
df.drop('date',inplace=True,axis=1)
```

```python
month_map = {
    '01': 'January', '02': 'February', '03': 'March', '04': 'April',
    '05': 'May', '06': 'June', '07': 'July', '08': 'August',
    '09': 'September', '10': 'October', '11': 'November', '12': 'December'
}


df['Month']=df['Date'].str.split('/').str[1].map(month_map)


```

```python
plt.figure(figsize=(20, 12))
plot_num = 1
for i in ['Month', 'Day_of_week', 'WeekStatus']:
    for j in ['Usage_kWh', 'CO2(tCO2)', 'NSM']:
        plt.subplot(3, 3, plot_num)
        sns.barplot(data=df.groupby(i).sum().reset_index(), x=i, y=j, color='steelblue')
        plt.title(f'{i} vs {j}')
        plt.xticks(rotation=45)
        plot_num += 1

plt.tight_layout()
plt.show()
```

```python
##This means the most of the power is going during weekdays , so commercial buildings might be taking up the electricity 
```

```python
df_grouped=df.groupby('Month', observed=True).sum().reset_index()
df_grouped.loc[df_grouped['Usage_kWh'].idxmax(), 'Month']
```

```python
df_grouped.loc[df_grouped['Usage_kWh'].idxmin(), 'Month']
```

```python
plt.figure(figsize=(15, 5))
for i, col in enumerate(['Usage_kWh', 'CO2(tCO2)', 'NSM'], 1):
    plt.subplot(1, 3, i)
    sns.histplot(data=df, x=col, bins=30, kde=True)
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 15))
plot_num = 1
for j in ['Usage_kWh', 'CO2(tCO2)', 'NSM','Load_Type']:
    plt.subplot(4, 1, plot_num)
    sns.histplot(data=df, x=j, bins=30, kde=True, color='steelblue')
    plt.title(f'Distribution of {j}')
    plot_num += 1
plt.tight_layout()
plt.show()
```

```python
df['hour'] = pd.to_datetime(df['time'], format='%H:%M').dt.hour
df['bin'] = (df['hour'] // 2) * 2
```

```python
agg = df.groupby('bin')['Usage_kWh'].sum().reset_index()
agg['label'] = agg['bin'].apply(lambda x: f"{x:02d}:00-{(x+2)%24:02d}:00")
```

```python
plt.figure(figsize=(11, 5))
bars = plt.bar(agg['label'], agg['Usage_kWh'], color='steelblue', edgecolor='black')

# Highlight the peak
peak_idx = agg['Usage_kWh'].idxmax()
bars[peak_idx].set_color('crimson')

plt.xlabel('Time range (2-hour bins)')
plt.ylabel('Total usage')
plt.title('Total usage per 2-hour interval')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(agg)
```
